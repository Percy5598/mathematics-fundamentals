### 1. Expected Value vs. Expected Utility

For discrete outcomes $x_i$ with probabilities $p_i$, the expected value is:
$$
\mathbb{E}[X] = \sum_i p_i x_i
$$

However, rational decision-makers often maximize **expected utility** rather than raw monetary expected value, applying a utility function $u(x)$ to reflect subjective well-being:
$$
EU(a) = \sum_i P(x_i \mid a) u(x_i)
$$

```python
import numpy as np

# Two hypothetical choices
outcomes_a = np.array([-100, 200])
probabilities_a = np.array([0.2, 0.8])

outcomes_b = np.array([40, 70])
probabilities_b = np.array([0.5, 0.5])

ev_a = probabilities_a @ outcomes_a
ev_b = probabilities_b @ outcomes_b

print(f"Expected Value Choice A: {ev_a:.2f}")
print(f"Expected Value Choice B: {ev_b:.2f}")
```

### 2. Risk Preferences and Utility Functions

A common utility function for wealth is square-root utility:
$$
u(w) = \sqrt{w}
$$
This function is **concave**, meaning each additional unit of wealth provides diminishing marginal utility (reflecting risk aversion).

```python
def utility(wealth):
    return np.sqrt(wealth)

wealth_a = np.array([100, 400])
prob_a = np.array([0.5, 0.5])

wealth_b = np.array([240, 240])
prob_b = np.array([0.5, 0.5])

print(f"Expected Wealth A:      {prob_a @ wealth_a:.2f}")
print(f"Expected Wealth B:      {prob_b @ wealth_b:.2f}")
print(f"Expected Utility A:     {prob_a @ utility(wealth_a):.4f}")
print(f"Expected Utility B:     {prob_b @ utility(wealth_b):.4f}")
```
