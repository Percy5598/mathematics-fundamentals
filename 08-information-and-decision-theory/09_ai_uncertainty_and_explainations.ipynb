### 1. Point Estimates versus Uncertainty

An AI model often outputs a single point estimate (e.g., a $70\%$ default probability). A more robust decision-making report includes calibration metrics, confidence intervals, and underlying data assumptions. 

> **Warning:** An AI model's internal confidence score is frequently miscalibrated and does not equate to a true frequentist probability.

```python
import numpy as np
import matplotlib.pyplot as plt

labels = ["Point Estimate", "Uncertainty Interval"]
point = 0.70
lower, upper = 0.55, 0.82

plt.figure(figsize=(8, 3))
plt.errorbar(
    [0], [point],
    yerr=[[point - lower], [upper - point]],
    fmt="o", color='darkcyan', capsize=8, markersize=8
)
plt.axhline(0.5, color='gray', linestyle="--", label="Decision Threshold (0.5)")
plt.xticks([0], ["Estimated Default Probability"])
plt.ylim(0, 1)
plt.ylabel("Probability Scale")
plt.title("Point Estimate with an Uncertainty Interval")
plt.legend(loc="upper right")
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.show()

print(f"Point Estimate:      {point}")
print(f"Uncertainty Interval: ({lower}, {upper})")
```

### 2. What Makes an Explanation Useful?

An effective explanation helps a decision-maker understand:
* Which specific features most strongly influenced the prediction.
* What input changes would alter the model's output (counterfactuals).
* The degree of uncertainty surrounding the prediction.
* Potential data limitations or domain boundary violations.
