### Case Study: AI-Assisted Investment Choice

An investor evaluates two actions (*Invest* vs. *Hold Cash*) across two market states (*Market Rises* vs. *Market Falls*). An AI system provides a predictive signal to guide the decision.

```python
import numpy as np
import pandas as pd

# Illustrative payoff matrix: actions by market state
actions = ["Invest", "Hold Cash"]
states = ["Market Rises", "Market Falls"]

payoffs = pd.DataFrame(
    [
        [120, -100],
        [0, 0],
    ],
    index=actions,
    columns=states,
)

print("Payoff Matrix ($):")
print(payoffs)

p_rise = 0.55
expected_payoffs = (
    p_rise * payoffs["Market Rises"] 
    + (1 - p_rise) * payoffs["Market Falls"]
)

print("\nExpected Payoffs (Prior p_rise = 0.55):")
print(expected_payoffs)
print(f"\nPreferred Action: {expected_payoffs.idxmax()}")
```

### Experimenting with an AI Signal

Suppose the AI signal is informative, modeled via Bayes' theorem:
$$
P(\text{Rise} \mid \text{Positive Signal}) = \frac{P(\text{Signal} \mid \text{Rise}) P(\text{Rise})}{P(\text{Signal})}
$$

```python
prior_rise = 0.55
p_positive_given_rise = 0.75
p_positive_given_fall = 0.30

p_positive = (
    p_positive_given_rise * prior_rise 
    + p_positive_given_fall * (1 - prior_rise)
)

posterior_rise = (p_positive_given_rise * prior_rise) / p_positive

print(f"Prior Probability of Market Rise:       {prior_rise:.4f}")
print(f"Posterior Probability (after signal):   {posterior_rise:.4f}")

updated_payoffs = (
    posterior_rise * payoffs["Market Rises"] 
    + (1 - posterior_rise) * payoffs["Market Falls"]
)

print("\nUpdated Expected Payoffs:")
print(updated_payoffs)
print(f"\nPreferred Action After Signal: {updated_payoffs.idxmax()}")
```

> **Disclaimer:** This simplified model is strictly for educational illustration and does not constitute financial advice. Real-world trading involves transaction costs, execution slippage, and structural market shifts.
