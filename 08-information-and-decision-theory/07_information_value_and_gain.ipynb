### 1. The Value of Information

The **Expected Value of Perfect Information (EVPI)** measures the maximum amount a decision-maker should be willing to pay for oracle-like information. It is defined as the expected payoff with perfect information minus the best expected payoff without it:
$$
\text{EVPI} = \mathbb{E}[\max_a U(a, S)] - \max_a \mathbb{E}[U(a, S)]
$$
Under standard utility models, EVPI is always non-negative.

```python
import numpy as np

# Rows are actions; columns are states of the world (e.g., [Strong, Weak])
payoffs = np.array([
    [100, -80],
    [50, 20],
    [20, 10],
])

probabilities = np.array([0.6, 0.4])

expected_by_action = payoffs @ probabilities
best_without_information = expected_by_action.max()

best_payoff_in_each_state = payoffs.max(axis=0)
expected_with_perfect_information = best_payoff_in_each_state @ probabilities

evpi = expected_with_perfect_information - best_without_information

print(f"Best expected payoff without information: {best_without_information:.2f}")
print(f"Expected payoff with perfect information: {expected_with_perfect_information:.2f}")
print(f"Calculated EVPI:                          {evpi:.2f}")
```

### 2. Information Gain vs. Decision Value

* **Information Gain:** Measures the reduction in uncertainty (entropy) after observing evidence, closely related to mutual information.
* **Practical Utility:** More information is not automatically valuable; to have positive economic value, data must be reliable, timely, relevant, and capable of altering an actual decision.
