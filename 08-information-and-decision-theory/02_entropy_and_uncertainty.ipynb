## Entropy and Uncertainty

### Learning Goals
* Compare entropy across different probability distributions (certain, concentrated, uniform).
* Understand why entropy measures outcome uncertainty rather than financial risk or downside loss.
* Appreciate the limitations of using information-theoretic metrics alone in investment or decision contexts.

---

### 1. Comparing Distributions

Entropy is maximized when all outcomes are equally likely (Uniform distribution) for a fixed number of possible states. It is lower when probability is concentrated on a small number of predictable outcomes.

```python
import numpy as np
import matplotlib.pyplot as plt

def entropy(p):
    p = np.asarray(p, dtype=float)
    p = p[p > 0]
    return -np.sum(p * np.log2(p))

distributions = {
    "Certain": [1.0, 0.0, 0.0],
    "Concentrated": [0.8, 0.1, 0.1],
    "Uniform": [1/3, 1/3, 1/3],
}

print("Entropy across distribution types:")
for name, p in distributions.items():
    print(f"  {name:<15} entropy = {entropy(p):.4f} bits")

names = list(distributions.keys())
values = [entropy(p) for p in distributions.values()]

plt.figure(figsize=(7, 4))
plt.bar(names, values, color=['skyblue', 'salmon', 'mediumseagreen'], edgecolor='black')
plt.ylabel("Entropy (bits)")
plt.title("Uncertainty Across Different Distributions")
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.show()
```

### 2. Entropy is Not the Same as Risk

> **Key Distinction:** Entropy describes uncertainty about *what* outcome will occur. Financial risk, by contrast, specifically concerns losses, downside exposure, volatility, or the probability of failing to meet a financial target. 

Two investments can possess identical outcome entropy while carrying vastly different financial consequences and risk profiles.
