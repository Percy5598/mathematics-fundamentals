## Risk, Uncertainty, and Ambiguity

### Learning Goals
* Distinguish between decision-making under *risk* (known probabilities) and *ambiguity* (unknown or disputed probabilities).
* Recognize how precise probability estimates can create false confidence.
* Understand how to communicate model uncertainty honestly.

---

### 1. Risk versus Ambiguity

* **Risk:** Outcomes occur with known or reliably estimable probabilities (e.g., tossing a fair coin).
* **Ambiguity (Knightian Uncertainty):** The underlying probabilities themselves are unknown, disputed, or impossible to measure reliably (e.g., predicting the exact timing or severity of a future financial crisis).

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

# Risk: Known probability p = 0.5
known_probability_trials = rng.binomial(1, 0.5, size=1000)

# Ambiguity: The true probability is unknown / drawn from a range
uncertain_p = rng.uniform(0.1, 0.9)
ambiguous_outcomes = rng.binomial(1, uncertain_p, size=1000)

print(f"Observed proportion of heads (Known Risk, true p=0.5): {known_probability_trials.mean():.3f}")
print(f"Underlying (unknown) probability in ambiguity example: {uncertain_p:.3f}")
print(f"Observed proportion of heads (Ambiguity): {ambiguous_outcomes.mean():.3f}")
```

### 2. Why the Distinction Matters

> **Core Insight:** Precise probability estimates can foster false confidence when models rest on weak foundations. Decision-makers must distinguish measurable random variability from fundamental uncertainty about model specification and parameter probabilities.
