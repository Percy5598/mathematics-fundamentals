### 3. Derivative Rules

### Core rules
* **Constant:** $\frac{d}{dx}(c) = 0$
* **Power:** $\frac{d}{dx}(x^n) = n x^{n-1}$
* **Constant multiple:** $\frac{d}{dx}[cf(x)] = c f'(x)$
* **Sum:** $(f + g)' = f' + g'$
* **Product:** $(fg)' = f'g + fg'$
* **Quotient:** $\left(\frac{f}{g}\right)' = \frac{f'g - fg'}{g^2}$, where $g \neq 0$
* **Chain:** $\frac{d}{dx}[f(g(x))] = f'(g(x))g'(x)$

### Examples
For f(x) = 3x⁴ − 2x² + 7, the derivative is 12x³ − 4x.

```python
import sympy as sp

x = sp.symbols('x')

functions = [
    3*x**4 - 2*x**2 + 7,
    (x**2 + 1)*(x**3 - 2),
    (3*x + 1)**5,
    sp.sin(x**2)
]

for f in functions:
    print("f(x) =", f)
    print("f'(x) =", sp.diff(f, x))
    print()
```

### Chain rule intuition

For $y = (3x+1)^5$, the outside function raises a quantity to the fifth power; the inside function is $3x + 1$. Differentiate the outside, keep the inside, then multiply by the inside derivative:

$$
y' = 5(3x+1)^4 \cdot 3 = 15(3x+1)^4
$$
