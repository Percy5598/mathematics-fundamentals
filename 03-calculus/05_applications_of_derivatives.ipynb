### 5. Applications of Derivatives

Derivatives help answer:
- How quickly is a quantity changing?
- Where is a function increasing or decreasing?
- Where are local maxima and minima?
- How can a business choose a quantity that maximizes profit?

### Critical points
Interior candidates for local maxima/minima often occur where f'(x)=0 or the derivative does not exist. They must be checked; not every critical point is an extremum.

```python
import sympy as sp
x = sp.symbols('x', real=True)
profit = -2*x**2 + 40*x - 100
marginal_profit = sp.diff(profit, x)
critical_points = sp.solve(sp.Eq(marginal_profit, 0), x)
print("Profit:", profit)
print("Marginal profit:", marginal_profit)
print("Critical point(s):", critical_points)
for point in critical_points:
    print(f"Profit at x={point}: {profit.subs(x, point)}")
```

### Second derivative test
If f'(a)=0 and f''(a)<0, the graph is locally concave down at a, indicating a local maximum. If f''(a)>0, it indicates a local minimum. If f''(a)=0, this test is inconclusive.

For the example profit function, the coefficient of x² is negative, so the parabola opens downward and its vertex is the maximum.


```python
import numpy as np
import matplotlib.pyplot as plt

q = np.linspace(0, 20, 200)
p = -2*q**2 + 40*q - 100
q_star = 10
p_star = -2*q_star**2 + 40*q_star - 100

plt.figure(figsize=(7, 4))
plt.plot(q, p, label="Profit(q)")
plt.scatter([q_star], [p_star], label=f"Maximum at q={q_star}")
plt.axhline(0, linewidth=1)
plt.xlabel("Quantity q"); plt.ylabel("Profit")
plt.title("A simple profit maximization problem")
plt.grid(True); plt.legend(); plt.show()
```

### Applications in ML and economics
- Gradient-based learning uses derivatives to update model parameters.
- Marginal cost and marginal revenue are derivatives of cost and revenue.
- Sensitivity analysis studies how outputs respond to small input changes.
