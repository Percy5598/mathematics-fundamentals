### A. Derivatives
1. Differentiate 4x³ − 5x + 2.
2. Differentiate (x² + 1)².
3. Find the second derivative of x³ − 2x².

### B. Optimization
4. Find the critical point of f(x)=−x²+8x−3 and determine whether it is a maximum.
5. A cost function is C(q)=q²+6q+40. Find marginal cost.

### C. Integration
6. Find an antiderivative of 6x² − 4.
7. Evaluate the definite integral of 2x from 0 to 5.

### D. Gradients
8. Find the gradient of f(x,y)=x²+xy+3y².
9. Explain gradient descent in plain language.

```python
import sympy as sp
x, y, q = sp.symbols('x y q')
problems = {
    "A1 derivative": sp.diff(4*x**3 - 5*x + 2, x),
    "A2 derivative": sp.diff((x**2 + 1)**2, x),
    "A3 second derivative": sp.diff(x**3 - 2*x**2, x, 2),
    "B4 critical point": sp.solve(sp.Eq(sp.diff(-x**2 + 8*x - 3, x), 0), x),
    "B5 marginal cost": sp.diff(q**2 + 6*q + 40, q),
    "C6 antiderivative": sp.integrate(6*x**2 - 4, x),
    "C7 definite integral": sp.integrate(2*x, (x, 0, 5)),
    "D8 gradient": [sp.diff(x**2 + x*y + 3*y**2, x), sp.diff(x**2 + x*y + 3*y**2, y)]
}
for question, answer in problems.items():
    print(f"{question}: {answer}")
```
