### 7. Definite Integrals and Area

A definite integral accumulates a signed quantity over an interval:
$$
\int_a^b f(x) \, dx
$$

Geometrically, it is net signed area: regions above the $x$-axis count positive and regions below count negative.

### Fundamental Theorem of Calculus

If $F'(x) = f(x)$, then
$$
\int_a^b f(x) \, dx = F(b) - F(a)
$$

Example:
$$
\int_0^2 x^2 \, dx = \left[ \frac{x^3}{3} \right]_0^2 = \frac{8}{3}
$$

```python
import sympy as sp
x = sp.symbols('x')
result = sp.integrate(x**2, (x, 0, 2))
print("Exact integral:", result)
print("Decimal:", float(result))
```

```python
import numpy as np
import matplotlib.pyplot as plt

x_vals = np.linspace(0, 2, 300)
y_vals = x_vals**2
plt.figure(figsize=(7, 4))
plt.plot(x_vals, y_vals, label="y = x²")
plt.fill_between(x_vals, y_vals, 0, alpha=0.25)
plt.xlabel("x"); plt.ylabel("y")
plt.title("Area under y = x² from 0 to 2")
plt.grid(True); plt.legend(); plt.show()
```

### Numerical approximation: trapezoidal rule

When an antiderivative is difficult or unavailable, approximate area with trapezoids:

$$
\int_a^b f(x) \, dx \approx \sum_{i=0}^{n-1} \frac{f(x_i) + f(x_{i+1})}{2} \Delta x
$$

```python
import numpy as np

x = np.linspace(0, 2, 1001)
y = x**2

approx = np.trapezoid(y, x)
print("Trapezoidal approximation:", approx)
print("Exact value:", 8/3)
```
