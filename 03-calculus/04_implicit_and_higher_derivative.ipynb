### 4. Implicit and Higher Derivatives

### Higher derivatives

The second derivative is the derivative of the derivative: $f''(x)$. It often describes acceleration or curvature.

### Example

If $s(t) = t^3 - 3t^2 + 2t$, then velocity is $s'(t) = 3t^2 - 6t + 2$ and acceleration is $s''(t) = 6t - 6$.

```python
import sympy as sp
t = sp.symbols('t')
s = t**3 - 3*t**2 + 2*t
v = sp.diff(s, t)
a = sp.diff(v, t)
print("Position:", s)
print("Velocity:", v)
print("Acceleration:", a)
print("At t=2: velocity =", v.subs(t, 2), ", acceleration =", a.subs(t, 2))
```

### Implicit differentiation

Sometimes $x$ and $y$ are connected by an equation rather than $y$ being isolated. For the circle $x^2 + y^2 = 25$, differentiate both sides with respect to $x$:

$$
2x + 2y \frac{dy}{dx} = 0 
\quad\Rightarrow\quad 
\frac{dy}{dx} = -\frac{x}{y}
$$

The factor $\frac{dy}{dx}$ appears because $y$ depends on $x$.

```python
x, y = sp.symbols('x y')
# Differentiate x^2 + y^2 = 25 implicitly with respect to x.
# SymPy can treat y as a function of x:
Y = sp.Function('y')(x)
equation = sp.Eq(x**2 + Y**2, 25)
implicit_derivative = sp.solve(sp.diff(equation.lhs, x), sp.diff(Y, x))[0]
print("dy/dx =", implicit_derivative)
```
