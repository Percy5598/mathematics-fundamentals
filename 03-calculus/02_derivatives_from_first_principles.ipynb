### 2. Derivatives from First Principles

### Intuition
Average rate of change is the slope between two points. A derivative is the slope we get as the second point moves infinitely close to the first.

### Definition
$$
f'(x)=\lim_{h\to 0}\frac{f(x+h)-f(x)}{h}
$$

This is called the difference quotient.

```python
def difference_quotient(x, h):
    f = lambda z: z**2
    return (f(x + h) - f(x)) / h

x = 3
for h in [1, 0.1, 0.01, 0.001, 0.0001]:
    print(f"h={h:>7}: quotient={difference_quotient(x, h):.6f}")
print("Expected derivative of x^2 at x=3: 6")
```

### Work it out for $f(x) = x^2$

$$
\frac{(x+h)^2 - x^2}{h} = \frac{x^2 + 2xh + h^2 - x^2}{h} = 2x + h
$$

Taking the limit as $h \to 0$ gives $f'(x) = 2x$.

```python
import numpy as np
import matplotlib.pyplot as plt

f = lambda x: x**2
x0 = 2
slope = 2*x0
x = np.linspace(0, 4, 200)

plt.figure(figsize=(7, 4))
plt.plot(x, f(x), label="f(x) = x²")
plt.plot(x, f(x0) + slope*(x-x0), "--", label=f"Tangent at x={x0}")
plt.scatter([x0], [f(x0)])
plt.xlabel("x"); plt.ylabel("y")
plt.title("Derivative as tangent slope")
plt.grid(True); plt.legend(); plt.show()
```
