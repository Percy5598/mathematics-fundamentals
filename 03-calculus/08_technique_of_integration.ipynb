### 8. Techniques of Integration

This lesson introduces two common methods: substitution and integration by parts.

### 1. Substitution

Substitution reverses the chain rule. If an integrand contains an inner expression and its derivative, set the inner expression equal to $u$.

For example:

$$
\int 2x(x^2+1)^3 \, dx
$$

Let $u = x^2 + 1$, so $du = 2x \, dx$. Then:

$$
\int u^3 \, du = \frac{u^4}{4} + C = \frac{(x^2+1)^4}{4} + C
$$
    

```python
import sympy as sp
x = sp.symbols('x')
expr = 2*x*(x**2 + 1)**3
F = sp.integrate(expr, x)
print("Antiderivative:", F)
print("Check:", sp.simplify(sp.diff(F, x) - expr))
```

## 2. Integration by parts

This comes from the product rule:

$$
\int u \, dv = uv - \int v \, du
$$

It is often useful for a product such as $x e^x$.

```python
expr = x * sp.exp(x)
print("Integral of x*exp(x):", sp.integrate(expr, x))
```
