### 1. Limits and Continuity

### Learning goals
- Understand a limit as the value a function approaches.
- Estimate limits numerically and graphically.
- Recognize continuity and common discontinuities.

### Intuition
A limit asks: **as x gets closer to a value, what does f(x) get closer to?** The function's value exactly at that point can be different—or even undefined.

For example, when x approaches 2, the expression `(x² - 4)/(x - 2)` approaches 4, even though the original expression is undefined at x = 2."""),


```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return (x**2 - 4) / (x - 2)

xs = np.array([1.9, 1.99, 1.999, 2.001, 2.01, 2.1])
print("x values:", xs)
print("f(x):", f(xs))
print("Expected limit as x -> 2: 4")
```

### Algebraic simplification

For $x \neq 2$,

$$
\frac{x^2 - 4}{x - 2} = \frac{(x - 2)(x + 2)}{x - 2} = x + 2
$$

Taking the limit as $x \to 2$ gives $4$. The limit describes nearby behavior, not necessarily the value at the point.


```python
x = np.linspace(1, 3, 500)
y = x + 2

plt.figure(figsize=(7, 4))
plt.plot(x, y, label="y = x + 2")
plt.scatter([2], [4], facecolors="white", edgecolors="red", s=90, label="Hole in original function")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Limit behavior near x = 2")
plt.grid(True)
plt.legend()
plt.show()
```

### Continuity
A function is continuous at x = a when:
1. f(a) is defined.
2. The limit as x approaches a exists.
3. The limit equals f(a).

### Applications
Limits support derivatives, instantaneous growth rates, marginal cost, and continuous-time models.
