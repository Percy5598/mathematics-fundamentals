### 6. Integrals and Antiderivatives

### Intuition
Differentiation measures local change. Integration can accumulate many small contributions, such as distance from velocity or total cost from marginal cost.

An antiderivative F satisfies F'(x)=f(x)

### Basic rules

$$
\int x^n \, dx = \frac{x^{n+1}}{n+1} + C, \quad n \neq -1
$$

$$
\int c \, dx = cx + C, \qquad \int (f + g) \, dx = \int f \, dx + \int g \, dx
$$

The constant $C$ is included because the derivative of any constant is zero.


```python
import sympy as sp
x = sp.symbols('x')
for f in [3*x**2, 2*x + 5, sp.cos(x), 1/x]:
    print("Integrand:", f, " -> antiderivative:", sp.integrate(f, x))
```

### Verify by differentiating
A reliable check is to differentiate the proposed antiderivative and see whether you recover the integrand.

```python
f = 3*x**2 + 4*x - 1
F = sp.integrate(f, x)
print("f(x) =", f)
print("F(x) =", F)
print("Derivative check:", sp.simplify(sp.diff(F, x) - f))
```
