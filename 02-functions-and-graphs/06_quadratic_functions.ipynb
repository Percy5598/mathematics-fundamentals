### 6. Quadratic Functions

A quadratic function has the form `f(x)=ax²+bx+c`, where `a != 0`. Its graph is a parabola.

- If `a > 0`, it opens upward and has a minimum.
- If `a < 0`, it opens downward and has a maximum.

The vertex is the turning point.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 300)
plt.figure(figsize=(7, 4))
plt.plot(x, x**2, label="y = x²")
plt.plot(x, -x**2, label="y = -x²")
plt.plot(x, x**2 - 4*x + 3, label="y = x² - 4x + 3")
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
plt.grid(True)
plt.legend()
plt.show()
```

### Vertex formula
For `f(x)=ax²+bx+c`, the x-coordinate of the vertex is `x_v=-b/(2a)`. The y-coordinate is `f(x_v)`.

The discriminant `D=b²-4ac` indicates the number of real x-intercepts: two if D>0, one repeated if D=0, and none if D<0.

```python
def quadratic_vertex(a, b, c):
    if a == 0:
        raise ValueError("a must be nonzero")
    x_vertex = -b / (2*a)
    y_vertex = a*x_vertex**2 + b*x_vertex + c
    return x_vertex, y_vertex

print(quadratic_vertex(1, -4, 3))  # (2, -1)
```
