### 2. Domain and Range

The **domain** is the set of allowed inputs. The **range** is the set of outputs the function actually produces.

Domain restrictions can arise from division by zero, square roots of negative values in the real numbers, or the context of a problem.

#### Examples
- `f(x)=x+2`: domain is all real numbers; range is all real numbers.
- `g(x)=1/x`: domain is all real numbers except 0; range is all real numbers except 0.
- `h(x)=sqrt(x)`: real-valued domain is `x >= 0`; range is `y >= 0`.
- `p(x)=x**2`: domain is all real numbers; range is `y >= 0`.


```python
import numpy as np

# reciprocal(0)  # Uncomment to see the domain error

def reciprocal(x):
    if x == 0:
        raise ValueError("x=0 is outside the domain")
    return 1/x

print(reciprocal(2))
```
