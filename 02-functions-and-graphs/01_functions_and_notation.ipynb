### 1. Functions and Notation

#### Learning objectives
- Understand a function as a rule mapping inputs to outputs.
- Read function notation such as `f(x)`.
- Evaluate functions manually and with Python.
- Distinguish a function from a general relationship.

#### What is a function?
A function assigns exactly one output to each valid input. If `f(x) = 2x + 3`, then the input `x=4` gives output `f(4)=11`.

The input is often called the independent variable; the output is often called the dependent variable. These names describe their roles in a model, not necessarily causation.


```python
def f(x):
    return 2*x + 3

for value in [0, 1, 4, -2]:
    print(f"f({value}) = {f(value)}")
```

### Function notation
`f(x)` means “the output of function f when its input is x.” It does not mean f multiplied by x.

For `f(x)=x**2-1`, `f(3)=3**2-1=8`.

```python
def f(x):
    return x**2 - 1

print(f(3))
print(f(-3))
```

### A function can be represented in several ways
- A verbal rule
- An equation
- A table of input-output pairs
- A graph
- A computer function

Different representations emphasize different features of the same relationship.

```python
inputs = list(range(-3, 4))
outputs = [f(x) for x in inputs]
list(zip(inputs, outputs))
```
