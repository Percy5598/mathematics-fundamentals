### 5. Linear Functions

A linear function in slope-intercept form is `f(x)=mx+b`.
- `m` is the slope.
- `b` is the y-intercept, the output when x=0.

In this form, the graph is a straight line. A linear model describes a constant rate of change.


```python
import numpy as np
import matplotlib.pyplot as plt

m, b = 2, 3
x = np.linspace(-5, 5, 100)
y = m*x + b

plt.figure(figsize=(7, 4))
plt.plot(x, y, label=f"y = {m}x + {b}")
plt.scatter([0], [b], label="y-intercept")
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()
```

### Find a line from two points
1. Calculate the slope `m=(y2-y1)/(x2-x1)`.
2. Substitute one point into `y=mx+b`.
3. Solve for `b`.

Example points `(1,5)` and `(3,9)` give slope 2. Using `(1,5)`, `5=2*1+b`, so `b=3`. The line is `y=2x+3`.

```python
x1, y1 = 1, 5
x2, y2 = 3, 9
m = (y2-y1)/(x2-x1)
b = y1 - m*x1
print("m =", m, "b =", b)
```

### Applications and caution
A linear model can represent constant costs, simple conversion rates, or a local approximation. Many real processes are not linear across their full range. A good fit in observed data does not guarantee valid extrapolation.
