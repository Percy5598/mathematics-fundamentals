### 4. Slope and Rate of Change

The slope between two distinct points `(x1,y1)` and `(x2,y2)` is:

`m = (y2 - y1) / (x2 - x1)`, provided `x2 != x1`.

It measures average change in y per unit change in x. In a straight-line graph, the slope is constant.
"""),

```python
x1, y1 = 1, 4
x2, y2 = 3, 10

slope = (y2 - y1) / (x2 - x1)
print("Slope:", slope)
```

### Interpret the sign
- Positive slope: y rises as x increases.
- Negative slope: y falls as x increases.
- Zero slope: y remains constant.
- Vertical line: slope is undefined because the change in x is zero.

Units matter. If y is euros and x is hours, slope is euros per hour.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)
plt.figure(figsize=(7, 4))
plt.plot(x, 2*x + 1, label="positive slope: 2x+1")
plt.plot(x, -x + 1, label="negative slope: -x+1")
plt.plot(x, np.ones_like(x)*2, label="zero slope: y=2")
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
plt.grid(True)
plt.legend()
plt.show()
```

### Average rate of change
For a function f from x=a to x=b:

`average rate = (f(b)-f(a))/(b-a)`, with `a != b`.

For nonlinear functions, this is the slope of the secant line between the two points. Instantaneous rate of change is a calculus concept covered later.


```python
def f(x):
    return x**2

a, b = 1, 3
average_rate = (f(b) - f(a)) / (b - a)
print(average_rate)  # (9 - 1) / 2 = 4
```
