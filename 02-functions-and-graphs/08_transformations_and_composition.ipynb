### 8. Transformations and Composition

Transformations modify a graph's position, orientation, or scale.

For a base function f(x):
- `f(x)+k` shifts the graph vertically by k.
- `f(x-h)` shifts the graph right by h.
- `a*f(x)` scales vertically by factor |a| and reflects across x-axis if a<0.
- `f(-x)` reflects across the y-axis.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 300)
plt.figure(figsize=(7, 4))
plt.plot(x, x**2, label="f(x)=x²")
plt.plot(x, x**2 + 3, label="f(x)+3")
plt.plot(x, (x-2)**2, label="f(x-2)")
plt.plot(x, -x**2, label="-f(x)")
plt.grid(True)
plt.show()
```

### Function composition
Composition means applying one function to the output of another: `(f∘g)(x)=f(g(x))`.

Order matters. In general, `f(g(x))` is not the same as `g(f(x))`.

```python
def f(x):
    return 2*x + 1

def g(x):
    return x**2

print("f(g(3)) =", f(g(3)))  # 19
print("g(f(3)) =", g(f(3)))  # 49
```
