### 9. Inverse Functions

An inverse function reverses the input-output mapping of a function. If `f(a)=b`, then its inverse satisfies `f⁻¹(b)=a`.

The notation `f⁻¹(x)` means inverse function, not `1/f(x)`.

### Find an inverse
Let `f(x)=2x+3`. Write `y=2x+3`, swap x and y to get `x=2y+3`, then solve for y: `y=(x-3)/2`. Thus `f⁻¹(x)=(x-3)/2`.

A function needs to be one-to-one on its domain to have an inverse function over its range.

```python
def f(x):
    return 2*x + 3

def inverse_f(y):
    return (y - 3)/2

for x in [-2, 0, 4]:
    print(x, inverse_f(f(x)))
```

### Horizontal line test
A graph is one-to-one if every horizontal line intersects it at most once. `y=x²` is not one-to-one over all real numbers, but it becomes one-to-one if its domain is restricted to `x>=0`.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 300)
plt.figure(figsize=(7, 4))
plt.plot(x, x**2, label="y=x²")
plt.plot(x, x, label="y=x")
plt.grid(True)
plt.legend()
plt.show()
```
