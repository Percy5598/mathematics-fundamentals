### 3. Coordinates and Graphs

A coordinate pair `(x, y)` identifies a point in a two-dimensional plane. A graph shows the relationship between inputs and outputs.

The horizontal axis is usually x; the vertical axis is usually y. If `y=f(x)`, each allowed x gives a point `(x, f(x))`.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 200)
y = x**2

plt.figure(figsize=(7, 4))
plt.plot(x, y, label="y = x²")
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of a quadratic function")
plt.grid(True)
plt.legend()
plt.show()
```

### Reading a graph
Ask:
- What is the output for a given input?
- Where does the graph cross either axis?
- Where is the function increasing or decreasing?
- Does it have a maximum or minimum?
- Which inputs are included in the domain?

A plotted curve is a visual aid. The equation is needed to establish exact values.

```python
# Plot several functions for comparison
x = np.linspace(-4, 4, 300)
plt.figure(figsize=(7, 4))
plt.plot(x, x, label="y = x")
plt.plot(x, x**2, label="y = x²")
plt.plot(x, 2*x + 1, label="y = 2x + 1")
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
plt.grid(True)
plt.legend()
plt.show()
```
