### 7. Exponential and Logarithmic Functions

An exponential function has the variable in the exponent, such as `f(x)=a*b**x`, where `b>0`. When `b>1`, it grows; when `0<b<1`, it decays.

A logarithmic function reverses an exponential: `y=log_b(x)` means `b**y=x`.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 4, 300)
plt.figure(figsize=(7, 4))
plt.plot(x, 2**x, label="y = 2^x")
plt.plot(x, (0.5)**x, label="y = (1/2)^x")
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
plt.grid(True)
plt.legend()
plt.show()
```

### Growth and decay
A compound-growth model can be written `V(t)=V0*(1+r)**t`, where r is the growth rate per period and t is the number of periods. The model assumes the rate remains constant and periods are comparable.

Logarithms are useful for solving for time or exponent values.

```python
import math

initial = 100
rate = 0.05
periods = 3
final = initial * (1 + rate)**periods
print(final)

# Solve for t in V = V0*(1+r)^t:
V0, V, r = 100, 150, 0.05
t = math.log(V / V0) / math.log(1 + r)
print("Periods to reach 150:", t)
```

### Domain and range
For `b**x` with b>0, the domain is all real x and the range is positive real values. For `log_b(x)`, the domain is x>0 and the range is all real numbers.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0.05, 5, 300)
plt.figure(figsize=(7, 4))
plt.plot(x, np.log(x), label="ln(x)")
plt.plot(x, np.log2(x), label="log₂(x)")
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
plt.grid(True)
plt.legend()
plt.show()
```
