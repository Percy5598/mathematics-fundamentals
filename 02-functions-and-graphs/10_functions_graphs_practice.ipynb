### Graphing challenge
Plot `y=x²`, `y=(x-2)²-1`, and `y=-0.5x+2` on the same axes. Label each curve and explain how the shapes differ.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 6, 300)
plt.figure(figsize=(8, 5))
plt.plot(x, x**2, label="y=x²")
plt.plot(x, (x-2)**2 - 1, label="y=(x-2)²-1")
plt.plot(x, -0.5*x + 2, label="y=-0.5x+2")
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Functions and transformations")
plt.grid(True)
plt.legend()
plt.show()
```
