### 1. Portfolio Weights and Mean-Variance Optimization

A portfolio weight describes the fraction of capital allocated to each asset. For a fully invested, long-only portfolio:
$$
\sum_i w_i = 1, \qquad w_i \geq 0
$$

* **Expected Portfolio Return:** $\mathbb{E}[R_p] = w^T\mu$
* **Portfolio Variance:** $\mathrm{Var}(R_p) = w^T\Sigma w$

where $\mu$ is the expected return vector and $\Sigma$ is the asset covariance matrix.

```python
import numpy as np
from scipy.optimize import minimize

expected_returns = np.array([0.08, 0.12, 0.06])

covariance = np.array([
    [0.040, 0.006, 0.004],
    [0.006, 0.090, 0.008],
    [0.004, 0.008, 0.025],
])

def portfolio_variance(weights):
    return weights @ covariance @ weights

# Constraints: weights sum to 1, expected return >= 8%
constraints = [
    {"type": "eq", "fun": lambda w: np.sum(w) - 1},
    {"type": "ineq", "fun": lambda w: w @ expected_returns - 0.08},
]

# Optimize to minimize variance
result = minimize(
    portfolio_variance,
    x0=np.array([1/3, 1/3, 1/3]),
    bounds=[(0, 1)] * 3,
    constraints=constraints,
)

if result.success:
    weights = result.x
    print(f"Optimal Weights: {weights}")
    print(f"Expected Return: {weights @ expected_returns:.4f}")
    print(f"Portfolio Variance: {portfolio_variance(weights):.4f}")
    print(f"Portfolio Volatility (SD): {np.sqrt(portfolio_variance(weights)):.4f}")
else:
    print(f"Optimization failed: {result.message}")
```

### 2. Interpretation and Limitations

> **Real-World Caution:** This is an educational illustration using hypothetical parameters. In practice, expected returns and covariance matrices are estimated with error. Real-world portfolio management must also account for transaction fees, taxes, liquidity constraints, and shifting market regimes.
