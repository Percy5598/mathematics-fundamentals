### 1. The Gradient Descent Update Rule

Gradient descent is an iterative optimization algorithm that repeatedly steps in the direction opposite to the gradient (the direction of steepest descent):
$$
x_{t+1} = x_t - \eta f'(x_t)
$$
where $\eta$ represents the **learning rate** (step size).

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return (x - 3)**2

def grad(x):
    return 2 * (x - 3)

# Gradient descent simulation
x = 0.0
learning_rate = 0.1
history = [x]

for _ in range(40):
    x = x - learning_rate * grad(x)
    history.append(x)

print(f"Final x after 40 iterations: {x:.4f}")
print(f"Final objective value: {f(x):.6f}")

plt.figure(figsize=(8, 4))
plt.plot(history, marker="o", color='teal', linewidth=2, label="Trajectory of $x$")
plt.axhline(3.0, color='crimson', linestyle="--", label="True Minimum ($x=3$)")
plt.xlabel("Iteration")
plt.ylabel("Decision Variable ($x$)")
plt.title("Gradient Descent Approaching the Minimum")
plt.legend()
plt.grid(linestyle=':', alpha=0.6)
plt.show()
```

### 2. Learning-Rate Trade-Offs

* **Too Small ($\eta$ is tiny):** Progress is agonizingly slow, requiring excessive iterations to reach convergence.
* **Too Large ($\eta$ is oversized):** Updates overshoot the minimum, causing oscillation or complete numerical divergence.
* **Scale Dependence:** The ideal learning rate depends on the steepness and scaling of the objective function.
