## Constraints and Feasible Regions

### Learning Goals

---

### 1. Decisions Have Limits

Real-world problems rarely allow unrestricted choices. A constrained optimization problem minimizes or maximizes an objective while ensuring that decisions respect boundaries like budgets, capacities, or physical laws.

Consider minimizing a distance squared subject to a bound:
$$
\min_x (x - 7)^2 \quad \text{subject to } 0 \leq x \leq 5
$$
The unconstrained minimum would occur at $x = 7$, but $7$ lies outside the **feasible region**. Consequently, the constrained minimum occurs at the boundary where $x = 5$.

```python
from scipy.optimize import minimize

def objective(x):
    return (x[0] - 7)**2

# Perform optimization with variable bounds
result = minimize(
    objective,
    x0=[2.0],              # Initial guess
    bounds=[(0, 5)],       # Feasible region bounds for x[0]
    method="L-BFGS-B"
)

print(f"Optimal Feasible x: {result.x[0]:.4f}")
print(f"Objective Function Value: {result.fun:.4f}")
print(f"Optimization Success: {result.success}")
```

### 2. Feasibility

> **Core Principle:** A solution is **feasible** if and only if it satisfies *every* constraint simultaneously. An optimal objective value is useless if the decision violates a mandatory boundary or regulatory limit.
