## Unconstrained Optimization

### Learning Goals
* Understand the concept of unconstrained optimization where decision variables are free from explicit limits.
* Use Python's `scipy.optimize.minimize_scalar` to find local and global minima efficiently.
* Recognize how numerical algorithms handle optimization problems and inspect solver outputs.

---

### 1. No Explicit Constraints

In unconstrained optimization, we search across the entire domain for the best value of a decision variable without boundary restrictions such as budget caps, storage limits, or non-negativity constraints.

Consider the quadratic objective function:
$$
f(x) = (x - 4)^2 + 1
$$
By inspection or calculus, its global minimum occurs at $x = 4$, yielding a minimum cost of $1$.

```python
import numpy as np
from scipy.optimize import minimize_scalar

def objective(x):
    return (x - 4)**2 + 1

# Perform scalar unconstrained optimization
result = minimize_scalar(objective)

print(f"Optimal x: {result.x:.4f}")
print(f"Minimum cost: {result.fun:.4f}")
print(f"Optimization Success: {result.success}")
```

### 2. Checking Results and Numerical Solvers

Numerical optimization algorithms rely on iterative steps, approximations, and stopping criteria (e.g., convergence tolerances). 
* Always inspect solver outputs (`result.success`, message flags) to ensure convergence.
* Where possible, compare numerical solutions with analytical math (calculus, first-order conditions where $f'(x) = 0$).
