### 1. What is Optimization?

Optimization means finding the best feasible choice according to a mathematical objective. 

**Common Examples:**
* A business maximizes profit or minimizes operational costs.
* An investor balances expected financial return against portfolio risk.
* A machine-learning model minimizes prediction error across a dataset.

We often formulate minimization mathematically as:
$$
x^* = \arg\min_x f(x)
$$
Here, $x$ is the **decision variable**, $f(x)$ is the **objective function** (cost), and $x^*$ is the optimal decision that achieves the minimum cost.

```python
import numpy as np
import matplotlib.pyplot as plt

def cost(x):
    return (x - 3) ** 2 + 2

x = np.linspace(-2, 8, 300)
y = cost(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y, color='teal', linewidth=2, label="Objective Function $f(x)$")
plt.scatter([3], [cost(3)], color='crimson', s=80, zorder=5, label=f"Optimal Point ($x^*=3$, Cost={cost(3)})")
plt.xlabel("Decision Variable ($x$)")
plt.ylabel("Cost ($f(x)$)")
plt.title("A Simple Quadratic Objective Function")
plt.legend()
plt.grid(linestyle=':', alpha=0.6)
plt.show()

print(f"Estimated Best x (via grid search): {x[np.argmin(y)]:.4f}")
print(f"Minimum cost at x=3: {cost(3)}")
```

### 2. Maximization versus Minimization

Optimization algorithms are frequently designed to find minimums by default. Fortunately, maximizing a utility or profit function $P(x)$ is mathematically equivalent to minimizing its negative:
$$
\max_x P(x) = \min_x [-P(x)]
$$
