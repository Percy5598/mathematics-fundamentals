## Convexity and Local versus Global Optima

### Learning Goals

---

### 1. Local and Global Minima

* **Local Minimum:** A decision $x^*$ is a local minimum if $f(x^*) \leq f(x)$ for all $x$ in a surrounding neighborhood.
* **Global Minimum:** A decision $x^*$ is a global minimum if $f(x^*) \leq f(x)$ across the *entire* feasible domain.

#### The Magic of Convexity
A function $f$ is **convex** if the line segment connecting any two points on its graph lies above or on the function curve:
$$
f(\alpha x + (1 - \alpha) y) \leq \alpha f(x) + (1 - \alpha) f(y) \quad \text{for all } \alpha \in [0, 1]
$$

> **Fundamental Theorem of Convex Optimization:** For a convex objective function defined on a convex feasible set, **every local minimum is also a global minimum**.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 400)
convex_y = x**2
nonconvex_y = x**4 - 2 * x**2 + 0.2

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Convex Plot
axes[0].plot(x, convex_y, color='teal', linewidth=2, label="$f(x) = x^2$")
axes[0].scatter([0], [0], color='red', s=60, zorder=5, label="Global Minimum")
axes[0].set_title("Convex Objective (Single Global Minimum)")
axes[0].set_xlabel("$x$")
axes[0].set_ylabel("$f(x)$")
axes[0].grid(linestyle=':', alpha=0.6)
axes[0].legend()

# Non-Convex Plot
axes[1].plot(x, nonconvex_y, color='darkorange', linewidth=2, label="$f(x) = x^4 - 2x^2 + 0.2$")
axes[1].scatter([-1, 1], [-0.8, -0.8], color='red', s=60, zorder=5, label="Global Minima")
axes[1].scatter([0], [0.2], color='purple', s=60, zorder=5, label="Local Maximum")
axes[1].set_title("Non-Convex Objective (Multiple Minima / Saddle)")
axes[1].set_xlabel("$x$")
axes[1].set_ylabel("$f(x)$")
axes[1].grid(linestyle=':', alpha=0.6)
axes[1].legend()

plt.tight_layout()
plt.show()
```

### 2. Why Convexity Matters

Convexity drastically simplifies numerical optimization:
* **Convex Problems:** Local search methods (e.g., gradient descent) are guaranteed to find the true global solution.
* **Non-Convex Problems:** Optimizers can easily get trapped in sub-optimal local minima, saddle points, or flat plateaus depending on initial starting conditions.
