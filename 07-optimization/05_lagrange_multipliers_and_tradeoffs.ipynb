### 1. Equality Constraints and the Lagrangian

When optimizing an objective function subject to strict equality constraints, the method of **Lagrange multipliers** converts a constrained problem into an unconstrained one by introducing auxiliary variables ($\lambda$).

For example, suppose we wish to maximize the product $f(x,y) = xy$ subject to a resource budget constraint:
$$
\max_{x,y} f(x,y) = xy \quad \text{subject to } x + y = 10
$$

We construct the Lagrangian function:
$$
\mathcal{L}(x,y,\lambda) = xy + \lambda(10 - x - y)
$$

The optimal value of the Lagrange multiplier ($\lambda$) represents the **shadow price**—the rate at which the optimal objective value changes if the constraint threshold is slightly relaxed.

```python
import numpy as np
from scipy.optimize import minimize

def objective(v):
    x, y = v
    return -(x * y)  # Minimize the negative product to maximize x * y

# Initial guess
x0 = [5.0, 5.0]

# Define equality constraint: x + y - 10 = 0
constraints = [
    {"type": "eq", "fun": lambda v: v[0] + v[1] - 10}
]

# Bounds ensuring non-negativity
bounds = [(0, None), (0, None)]

result = minimize(
    objective,
    x0,
    constraints=constraints,
    bounds=bounds,
    method="SLSQP"
)

print(f"Optimal x, y: {result.x}")
print(f"Maximized Product (xy): {-result.fun:.4f}")
print(f"Constraint Sum (x + y): {result.x.sum():.4f}")
print(f"Optimization Success: {result.success}")
```
