### Learning Goals

---

### 1. Optimizing a Two-Variable Objective

When objective functions involve multiple decision variables and cannot be easily solved by hand, numerical algorithms find local or global minima iteratively.

Consider a multi-variable quadratic function:
$$
f(x, y) = (x - 2)^2 + 3(y + 1)^2
$$

```python
import numpy as np
from scipy.optimize import minimize

def objective(v):
    x, y = v
    return (x - 2)**2 + 3*(y + 1)**2

# Run unconstrained optimization from initial guess [0, 0]
result = minimize(objective, x0=[0.0, 0.0])

print(f"Optimal Solution (x, y): {result.x}")
print(f"Minimum Objective Value: {result.fun:.4f}")
print(f"Solver Success: {result.success}")
print(f"Status Message: {result.message}")
```

### 2. Validating Numerical Results

> **Validation Checklist:**
> * **Success Flag:** Did the algorithm report convergence (`result.success == True`)?
> * **Plausibility:** Is the final objective value logically consistent with the problem domain?
> * **Constraints:** Are all boundary and equality constraints satisfied?
> * **Analytical Check:** Does the result agree with hand calculations or first-order conditions ($f'(x) = 0$)?
> * **Sensitivity:** Does the solver reach the same solution from multiple distinct starting points?
