## The Mathematics of Machine Learning

### Learning Goals

* Distinguish between model features ($x$), adjustable parameters ($\theta$), and loss functions ($L(\theta)$).
* Implement and visualize simple linear models with varying weights and biases.

---

### 1. A Machine-Learning Model is a Mathematical Function

A machine learning model maps input features to a prediction via a parametrized function:
$$
\hat{y} = f_\theta(x)
$$
where $x$ represents the input data, $\theta$ represents adjustable model parameters, and $\hat{y}$ is the resulting prediction. 

Training a model means finding parameters that optimize performance across data by minimizing a loss function:
$$
\theta^* = \arg\min_\theta L(\theta)
$$

### 2. A Simple Linear Model

Consider a univariate linear model:
$$
\hat{y} = wx + b
$$
The weight $w$ controls the slope (sensitivity to input changes), and $b$ controls the vertical intercept. Training attempts to find useful values for both.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 100)

weights = [-2.0, 0.5, 2.0]
bias = 1.0

plt.figure(figsize=(8, 4))
for w in weights:
    prediction = w * x + bias
    plt.plot(x, prediction, linewidth=2, label=f"$w = {w}, b = {bias}$")

plt.xlabel("Input Feature ($x$)")
plt.ylabel("Model Prediction ($\hat{y}$)")
plt.title("A Linear Model with Different Weights ($w$)")
plt.legend()
plt.grid(linestyle=':', alpha=0.6)
plt.show()
```

### 3. Why Training is Necessary

A model with arbitrary parameters makes poor predictions. Training adjusts parameters using training examples and an optimization algorithm. 

The training process relies on three central ingredients:
1. **A model architecture** that generates predictions.
2. **A loss function** that evaluates prediction errors.
3. **An optimization algorithm** (e.g., gradient descent) that updates parameters.
