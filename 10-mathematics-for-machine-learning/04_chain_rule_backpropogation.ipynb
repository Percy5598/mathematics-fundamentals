### 1. The Calculus Chain Rule

Neural networks are compositions of nested mathematical functions. For composite functions $y = f(u)$ where $u = g(x)$, the derivative of $y$ with respect to $x$ is given by the chain rule:
$$
\frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx}
$$

Consider a single-neuron forward pass:
$$
z = wx + b, \qquad \hat{y} = z^2
$$

Applying the chain rule to find the sensitivity of the prediction $\hat{y}$ to the weight parameter $w$:
$$
\frac{d\hat{y}}{dw} = \frac{d\hat{y}}{dz} \cdot \frac{dz}{dw} = (2z) \cdot (x) = 2zx
$$

```python
import numpy as np

x = 2.0
w = 3.0
b = 1.0

def model(weight):
    z = weight * x + b
    return z**2

# Analytical derivative via Chain Rule
z = w * x + b
analytical = 2 * z * x

# Numerical differentiation via Central Finite Difference
h = 1e-5
numerical = (model(w + h) - model(w - h)) / (2 * h)

print(f"Analytical Derivative (Chain Rule): {analytical:.6f}")
print(f"Numerical Derivative (Finite Difference): {numerical:.6f}")
print(f"Close Match? {np.isclose(analytical, numerical)}")
```

### 2. Backpropagation in Neural Networks

For deep neural networks with multiple hidden layers and activation functions, backpropagation applies the chain rule systematically in reverse—transmitting error gradients from the final loss function back to every individual weight and bias parameter.

> **Computational Efficiency:** Computing gradients analytically by hand for millions of parameters is intractable. Automatic differentiation engines construct computational graphs to evaluate gradients efficiently in time proportional to forward pass computation.
