### 1. Distance and Similarity Metrics

* **Euclidean Distance:** Measures geometric straight-line separation in vector space:
  $$
  d(x, y) = \sqrt{\sum_j (x_j - y_j)^2}
  $$
  Large-scale features disproportionately dominate Euclidean distance unless features are standardized.

* **Cosine Similarity:** Measures the directional angle between two vectors, ignoring magnitude:
  $$
  \cos(x, y) = \frac{x^T y}{\|x\| \|y\|}
  $$

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics.pairwise import (
    euclidean_distances,
    cosine_similarity,
    rbf_kernel,
)
from sklearn.preprocessing import StandardScaler

a = np.array([[1, 2], [2, 4], [-1, -2], [1, -2]])

print("Euclidean Distances:")
print(np.round(euclidean_distances(a), 2))

print("\nCosine Similarity:")
print(np.round(cosine_similarity(a), 2))

print("\nRBF Kernel Similarity (gamma=0.2):")
print(np.round(rbf_kernel(a, gamma=0.2), 2))

scaled = StandardScaler().fit_transform(
    np.array([[1, 1000], [2, 1100], [3, 3000]])
)
print("\nStandardized Features (Scaling Demonstration):")
print(np.round(scaled, 2))
```

### 2. Kernel Methods and Non-Linearity

The **Radial Basis Function (RBF) kernel** converts distance into a bounded similarity score:
$$
K(x, y) = \exp(-\gamma \|x - y\|^2)
$$
Kernel methods allow algorithms (such as SVMs) to learn complex non-linear decision boundaries in high-dimensional feature spaces without explicitly computing expensive coordinate transformations.
