##### Linear Equations

An equation is a statement that two expressions are equal. Solving an equation means finding values of the variable that make it true.

The central rule: perform the same valid operation on both sides to preserve equality.

##### Example: solve `3x + 5 = 20`

1. Subtract 5 from both sides: `3x = 15`.
2. Divide both sides by 3: `x = 5`.
3. Check: `3*5 + 5 = 20`.

```python
x = 5
print(3*x + 5)
assert 3*x + 5 == 20
```

##### General form

For `ax + b = c`, when `a != 0`, the solution is `x = (c-b)/a`.

Some equations have no solution or infinitely many solutions. For example, `x+1=x+2` has no solution; `x+1=x+1` is true for every x.

```python
import sympy as sp

x = sp.symbols("x")
print(sp.solve(sp.Eq(3*x + 5, 20), x))
print(sp.solve(sp.Eq(x + 1, x + 2), x))
print(sp.solve(sp.Eq(x + 1, x + 1), x))
```

##### Rearranging formulas

Suppose `y = mx + b`. To isolate x, subtract b and divide by m (provided `m != 0`): `x = (y-b)/m`.

This is useful in finance, statistics, and machine learning when solving for an unknown quantity.
