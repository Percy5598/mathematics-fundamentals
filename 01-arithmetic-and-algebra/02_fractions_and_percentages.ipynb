#### 2. Fractions, Decimals, and Percentages

##### Core ideas
A fraction `a/b` represents `a` divided by `b`, where `b` cannot be zero. A decimal is another way to represent a number. A percentage means “per hundred.”

Examples: `1/2 = 0.5 = 50%`.

```python
from fractions import Fraction

print(Fraction(1, 2))
print(Fraction(1, 3) + Fraction(1, 6))  # 1/2
print(float(Fraction(3, 8)))             # 0.375
```

#### Fraction operations

For addition/subtraction, use a common denominator:

`1/4 + 1/2 = 1/4 + 2/4 = 3/4`

For multiplication, multiply numerators and denominators.

For division by a nonzero fraction, multiply by its reciprocal:

`(2/3) / (4/5) = (2/3) * (5/4) = 5/6`

```python
a = Fraction(1, 4) + Fraction(1, 2)
b = Fraction(2, 3) * Fraction(4, 5)
c = Fraction(2, 3) / Fraction(4, 5)
print(a, b, c)
```

##### Percentages

`percentage = part / whole * 100`

`new_value = old_value * (1 + rate)` for an increase, where rate is decimal form.

`new_value = old_value * (1 - rate)` for a decrease.

A 20% increase followed by a 20% decrease does not return to the starting value because the second percentage is calculated from the changed value.

```python
old_price = 100
increased = old_price * 1.20
after_decrease = increased * 0.80

print("After 20% increase:", increased)
print("Then after 20% decrease:", after_decrease)
print("Overall change (%):", (after_decrease / old_price - 1) * 100)

```

##### Percentage change

`percentage_change = (new - old) / old * 100`

This formula requires a meaningful nonzero starting value. If the old value is zero, the percentage change is undefined.

```python
def percentage_change(old, new):
    if old == 0:
        raise ValueError("Percentage change is undefined when the old value is zero.")
    return (new - old) / old * 100

print(percentage_change(100, 120))  # 20%
print(percentage_change(100, 80))   # -20%
print(percentage_change(80, 100))   # 25%
```

#####  Ratios and Proportions

A ratio compares quantities. The ratio 2:3 means two parts for every three parts. A proportion states that two ratios are equal.

Ratios are sensitive to units: compare like with like, and convert units when necessary.

```python
# A recipe uses 2 cups of rice for 3 cups of water.
rice = 2
water = 3
scale_factor = 4

print("Rice for scaled recipe:", rice * scale_factor)
print("Water for scaled recipe:", water * scale_factor)
```

##### Direct proportion

If `y = k*x`, then y changes by the same factor as x. The constant `k = y/x` is the proportionality constant when `x != 0`.

##### Inverse proportion

If `y = k/x`, multiplying x by a factor divides y by that factor, for nonzero x.

```python
# If 4 workers complete a fixed task in 6 hours,
# idealized inverse proportion predicts 8 workers need 3 hours.
workers_1, hours_1 = 4, 6
workers_2 = 8
hours_2 = workers_1 * hours_1 / workers_2
print(hours_2)
```

##### Rates

A rate compares quantities with different units, such as kilometers per hour, euros per item, or revenue per customer. Always write units when interpreting a rate.
