##### Polynomials and Factoring

A polynomial is a sum of terms with nonnegative integer powers of variables, such as `3x**2 - 2x + 1`.

The degree is the highest exponent with a nonzero coefficient. The degree of `3x**2 - 2x + 1` is 2.

##### Expanding and factoring

Expanding multiplies terms out: `(x+2)(x+3) = x**2 + 5x + 6`.

Factoring reverses the process: `x**2 + 5x + 6 = (x+2)(x+3)`.


```python
import sympy as sp

x = sp.symbols("x")
expr = (x + 2) * (x + 3)
print("Expanded:", sp.expand(expr))
print("Factored:", sp.factor(sp.expand(expr)))
```

##### Common identities

- `(a+b)**2 = a**2 + 2ab + b**2`
- `(a-b)**2 = a**2 - 2ab + b**2`
- `(a+b)(a-b) = a**2 - b**2`

A common mistake is to omit the middle term in a squared binomial.

```python
a, b = sp.symbols("a b")
print(sp.expand((a + b)**2))
print(sp.expand((a - b)**2))
print(sp.expand((a + b)*(a - b)))
```

```python
x = sp.symbols("x")
print(sp.expand((x + 4)*(x + 1)))
print(sp.factor(x**2 + 7*x + 12))
print(sp.expand((x - 3)**2))
```
