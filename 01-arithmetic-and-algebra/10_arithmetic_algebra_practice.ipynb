##### Questions

1. Evaluate `18 - 3 * (2 + 1)`.
2. Convert `3/5` to a decimal and percentage.
3. A price moves from 50 to 65. Find the percentage change.
4. Simplify `4x + 7 - 2x + 3`.
5. Solve `3x - 8 = 13`.
6. Solve `-2x <= 10`.
7. Factor `x**2 + 9x + 20`.
8. Solve `x**2 - 7x + 12 = 0`.
9. Calculate `log_2(32)`.
10. A value increases by 10% and then decreases by 10%. Find the overall percentage change.


```python
from fractions import Fraction
import math
import sympy as sp

x = sp.symbols("x")
solutions = {
    1: 18 - 3 * (2 + 1),
    2: (float(Fraction(3, 5)), float(Fraction(3, 5)) * 100),
    3: (65 - 50) / 50 * 100,
    4: sp.simplify(4*x + 7 - 2*x + 3),
    5: sp.solve(sp.Eq(3*x - 8, 13), x),
    6: sp.solve_univariate_inequality(-2*x <= 10, x),
    7: sp.factor(x**2 + 9*x + 20),
    8: sp.solve(x**2 - 7*x + 12, x),
    9: math.log(32, 2),
    10: ((1.10 * 0.90) - 1) * 100,}
```
