##### 5. Algebraic Expressions and Simplification

A variable is a symbol representing a value. An expression combines numbers, variables, and operations. An equation asserts that two expressions are equal.

Example: `3x + 2` is an expression; `3x + 2 = 11` is an equation.

##### Combining like terms

`3x + 2x = 5x`, but `3x + 2` cannot be combined into one like term because one term contains x and the other is constant.

##### Distributive property

`a(b+c) = ab + ac`

Thus `3(x+4) = 3x + 12`.

```python
import sympy as sp

x, y = sp.symbols("x y")
expr = 3*(x + 4) + 2*x - 5
print("Expanded:", sp.expand(expr))
print("Simplified:", sp.simplify(expr))
```

##### Substitution

To evaluate `2x + 3` at `x=4`, substitute 4 for x: `2*4 + 3 = 11`.
