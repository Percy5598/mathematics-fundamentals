##### Number systems
- **Natural numbers:** counting numbers; conventions differ on whether zero is included.
- **Integers:** whole numbers including negatives, zero, and positives.
- **Rational numbers:** numbers expressible as a ratio of integers, such as `3/4` or `-2`.
- **Irrational numbers:** real numbers that cannot be written as a ratio of integers, such as `sqrt(2)` and `pi`.
- **Real numbers:** rational and irrational numbers together.

##### Four operations

Addition combines quantities: `7 + 3 = 10`.

Subtraction finds a difference: `7 - 10 = -3`.

Multiplication scales a quantity: `4 * 3 = 12`.

Division splits a quantity or forms a ratio: `12 / 4 = 3`.

Division by zero is undefined.


```python
# Basic Operations
a = 12
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
```

##### Negative numbers

On a number line, moving right increases a value and moving left decreases it.

`5 - 8 = 5 + (-8) = -3`

For multiplication:
- positive × positive = positive
- negative × negative = positive
- positive × negative = negative
- negative × positive = negative

The sign rules are consistent with the distributive property.

```python
print(-7 + 12)       # 5
print(5 - 8)         # -3
print((-3) * (-4))   # 12
print(17 % 5)        # 2
print(-7 // 2)       # -4: floor division rounds down
```

```python
old_price = 100
new_price = 120

absolute_change = new_price - old_price

percentage_change = (
    absolute_change / old_price
) * 100

print(f"Absolute change: €{absolute_change}")
print(f"Percentage change: {percentage_change:.2f}%")
```

##### Order of operations

Use parentheses first, then exponents, then multiplication/division from left to right, then addition/subtraction from left to right.

For example: `2 + 3 * 4 = 14`, while `(2 + 3) * 4 = 20`.

Multiplication and division have equal precedence; so do addition and subtraction.

```python
print(2 + 3 * 4)       # 14
print((2 + 3) * 4)     # 20
print(12 / 3 * 2)      # 8.0
print(2 ** 3 + 1)      # 9
```

##### Important properties

- Commutative: `a + b = b + a`; `a * b = b * a`
- Associative: `(a + b) + c = a + (b + c)`
- Distributive: `a * (b + c) = a*b + a*c`
- Additive identity: `a + 0 = a`
- Multiplicative identity: `a * 1 = a`

Subtraction and division are not generally commutative.

```python
a, b, c = 3, 4, 5

assert a + b == b + a
assert a * b == b * a
assert (a + b) + c == a + (b + c)
assert a * (b + c) == a*b + a*c

print("Properties verified.")
```
