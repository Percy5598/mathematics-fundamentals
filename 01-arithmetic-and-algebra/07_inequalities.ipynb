##### Inequalities

Inequalities compare values:
- `<` less than
- `>` greater than
- `<=` less than or equal to
- `>=` greater than or equal to

Solving an inequality finds all values that satisfy the comparison.

##### Solving inequalities

Solve `2x + 3 < 11`:
1. Subtract 3: `2x < 8`.
2. Divide by positive 2: `x < 4`.

**Important:** multiplying or dividing both sides by a negative number reverses the inequality sign. Example: `-2x < 8` gives `x > -4`.


```python
import sympy as sp

x = sp.symbols("x", real=True)
print(sp.solve_univariate_inequality(2*x + 3 < 11, x))
print(sp.solve_univariate_inequality(-2*x < 8, x))
```

##### Inequalities in practice
Thresholds appear in credit risk, classification, budgeting, and quality control. For example, a decision rule might flag a score when `risk_score >= threshold`. The mathematical inequality alone does not establish whether the threshold is appropriate; that requires evidence and context.
