#### 3. Powers, Roots, and Logarithms

##### Powers
An exponent tells you how many times a base is multiplied by itself: `2**3 = 2*2*2 = 8`.

For nonzero `a`: `a**0 = 1`; `a**(-n) = 1/(a**n)`.

```python
print(2**5)
print(3**0)
print(2**-3)
print(16**0.5)
```

##### Exponent rules

For suitable nonzero bases:
- `a**m * a**n = a**(m+n)`
- `a**m / a**n = a**(m-n)`
- `(a**m)**n = a**(m*n)`
- `(a*b)**n = a**n * b**n`

A root reverses a power: `sqrt(25) = 5` for the principal square root.

```python
import math

print(math.sqrt(25))
print(math.sqrt(2))
print(27 ** (1/3))
```

##### Logarithms

`log_b(x) = y` means `b**y = x`, with `b > 0`, `b != 1`, and `x > 0`.

Example: `log_2(8) = 3` because `2**3 = 8`.

Natural logarithm `ln(x)` uses base `e`. In Python, `math.log(x)` is the natural logarithm by default.


```python
import math

print(math.log(8, 2))  # 3
print(math.log(100, 10))  # 2
print(math.log(math.e))  # 1
```
