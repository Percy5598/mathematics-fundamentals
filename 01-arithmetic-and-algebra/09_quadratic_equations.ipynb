##### Quadratic Equations

A quadratic equation has the form `ax**2 + bx + c = 0`, where `a != 0`.

Its solutions are roots of the quadratic. Depending on the coefficients, it may have two distinct real roots, one repeated real root, or two complex roots.

##### Quadratic formula

For `ax**2 + bx + c = 0`:

`x = (-b ± sqrt(b**2 - 4ac)) / (2a)`

The discriminant `D = b**2 - 4ac` tells us about the roots:
- `D > 0`: two distinct real roots
- `D = 0`: one repeated real root
- `D < 0`: no real roots, but two complex roots


```python
import math

def quadratic_roots(a, b, c):
    if a == 0:
        raise ValueError("a must not be zero for a quadratic equation")
    discriminant = b*b - 4*a*c
    if discriminant >= 0:
        return ((-b + math.sqrt(discriminant))/(2*a),
                (-b - math.sqrt(discriminant))/(2*a))
    real = -b/(2*a)
    imag = math.sqrt(-discriminant)/(2*a)
    return (complex(real, imag), complex(real, -imag))

print(quadratic_roots(1, -5, 6))  # (3, 2)
print(quadratic_roots(1, 2, 1))   # (-1, -1)
print(quadratic_roots(1, 0, 1))   # complex roots
```

##### Solve with SymPy
Symbolic tools are useful for checking algebra, but you should still learn how to solve equations manually.


```python
import sympy as sp

x = sp.symbols("x")
print(sp.solve(sp.Eq(x**2 - 5*x + 6, 0), x))
print(sp.factor(x**2 - 5*x + 6))
```
