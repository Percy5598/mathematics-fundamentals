### 1. The One-Sample $t$-Test

Hypothesis testing provides a formal framework for deciding whether observed data contradict a baseline assumption (the null hypothesis). For a population mean $\mu$, we typically test:
* **Null Hypothesis ($H_0$):** $\mu = \mu_0$ (no effect or no difference from baseline)
* **Alternative Hypothesis ($H_1$):** $\mu \neq \mu_0$ (two-sided test indicating a difference)

The test statistic for a one-sample $t$-test is:
$$
t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}
$$

```python
import numpy as np
from scipy import stats

# Sample daily financial returns
returns = np.array([0.4, 1.2, -0.3, 0.8, 1.0, -0.1, 0.6, 0.2, 1.4, -0.5])

# Perform a two-sided one-sample t-test against a population mean of 0
result = stats.ttest_1samp(returns, popmean=0)

print(f"Sample Mean: {returns.mean():.4f}")
print(result)
```

### 2. Decision Rules and Significance Levels

To make an objective decision, we compare the resulting $p$-value to a pre-chosen significance level ($\alpha$, commonly set to $0.05$):
* If $p < \alpha$, we reject the null hypothesis in favor of the alternative.
* If $p \ge \alpha$, we fail to reject the null hypothesis.

```python
alpha = 0.05

decision = "Reject H0 (Statistically Significant)" if result.pvalue < alpha else "Do not reject H0 (Not Statistically Significant)"
print(f"Alpha level: {alpha}")
print(f"P-value: {result.pvalue:.4f}")
print(f"Decision: {decision}")
```

### Interpretation and Error Types

> **Accurate Definition of a $P$-value:** A $p$-value is the probability, *assuming the null hypothesis and model assumptions are true*, of obtaining a test statistic at least as extreme as the one observed. It is **not** $P(H_0 \text{ is true})$, nor is it the probability that the result happened purely "by chance."

* **Decision Errors:**
  * **Type I Error ($\alpha$):** Rejecting a true null hypothesis (a false positive).
  * **Type II Error ($\beta$):** Failing to reject a false null hypothesis (a false negative).
* **Statistical vs. Practical Importance:** A statistically significant result simply means the effect is unlikely to be zero by random sampling variation; it does *not* automatically imply the effect size is large or operationally useful.
