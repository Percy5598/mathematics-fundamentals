### Addition Rule and Complements

* **Complement Rule:** The complement of an event $A$ (denoted $A^c$ or $\text{not } A$) represents the event where $A$ does not occur:
  $$
  P(A^c) = 1 - P(A)
  $$

* **General Addition Rule:** For any two events $A$ and $B$, the probability that at least one of them occurs is:
  $$
  P(A \cup B) = P(A) + P(B) - P(A \cap B)
  $$

* **Mutually Exclusive Events:** If $A$ and $B$ cannot occur simultaneously, they are mutually exclusive, meaning $P(A \cap B) = 0$ and the rule simplifies to:
  $$
  P(A \cup B) = P(A) + P(B)
  $$
