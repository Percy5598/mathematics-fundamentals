### Joint Distributions, Correlation, and Dependence

When analyzing multiple random variables simultaneously, their probabilistic interactions require specialized concepts:

* **Joint Distributions:** Describe the simultaneous behavior and probabilities of two or more random variables together.
* **Marginal Distributions:** Describe the probability distribution of a single subset of variables within a joint distribution, isolated from the others.
* **Correlation:** Measures the strength and direction of the *linear* association between variables. 

> **Important Caveats:**
> * **Zero Correlation $\neq$ Independence:** A correlation of zero indicates no *linear* relationship, but complex nonlinear dependencies can still exist between the variables.
> * **Correlation vs. Causation:** Tracking how variables vary together establishes association, not a causal mechanism.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(19)
x = rng.uniform(-2, 2, 1000)
y = x**2 + rng.normal(0, 0.25, size=len(x))
print("Correlation for nonlinear relationship:", np.corrcoef(x, y)[0, 1])
plt.scatter(x, y, alpha=0.35, s=15)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("A nonlinear relationship can have low correlation")
plt.grid(True, alpha=0.3)
plt.show()
```
