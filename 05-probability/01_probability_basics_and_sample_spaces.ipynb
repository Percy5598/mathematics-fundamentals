### Probability Basics and Sample Spaces

Probability provides a mathematical framework for quantifying uncertainty. Every probability $P(A)$ for an event $A$ satisfies:
$$
0 \le P(A) \le 1
$$
where $0$ indicates an impossible event and $1$ indicates absolute certainty.

* **Sample Space ($\Omega$):** The complete set of all possible outcomes of a random experiment.
* **Event ($A$):** Any subset of the sample space ($\Omega$).

For a sample space consisting of equally likely outcomes, the probability of an event $A$ is calculated as:
$$
P(A) = \frac{|A|}{|\Omega|} = \frac{\text{number of favorable outcomes in } A}{\text{total number of possible outcomes}}
$$

```python
import numpy as np

# Fair six-sided die
sample_space = np.arange(1, 7)
event_even = sample_space[sample_space % 2 == 0]
print("Sample space:", sample_space)
print("Even outcomes:", event_even)
print("P(even), equally likely outcomes:", len(event_even) / len(sample_space))

# Simulate many rolls
rng = np.random.default_rng(42)
rolls = rng.integers(1, 7, size=10_000)
print("Simulated proportion even:", np.mean(rolls % 2 == 0))
```
