### Application: Simulating the Law of Large Numbers and Bayes' Theorem

Probability forms the mathematical backbone for statistical inference, risk modeling, decision-making, and machine learning.


```python
import numpy as np

# Monte Carlo estimate of the probability that the sum of two fair dice is 7
rng = np.random.default_rng(99)
n = 200_000
die1 = rng.integers(1, 7, size=n)
die2 = rng.integers(1, 7, size=n)
estimate = np.mean(die1 + die2 == 7)
theoretical = 6 / 36
print("Monte Carlo estimate:", estimate)
print("Theoretical probability:", theoretical)

# A small Bayes calculation
p_a = 0.02
p_pos_given_a = 0.90
p_pos_given_not_a = 0.05
p_pos = p_pos_given_a*p_a + p_pos_given_not_a*(1-p_a)
print("P(A | positive):", p_pos_given_a*p_a/p_pos)
```
