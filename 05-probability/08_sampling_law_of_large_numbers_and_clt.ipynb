### Sampling, Law of Large Numbers, and Central Limit Theorem

Statistical inference relies heavily on understanding how sample statistics behave as we collect more data. Two foundational theorems govern this behavior:

* **Law of Large Numbers (LLN):** States that as the sample size $n$ increases, the sample average $\bar{X}_n$ converges toward the true population expected value $\mu = E[X]$. In practical terms: *more data yields more stable and accurate estimates.*
  
* **Central Limit Theorem (CLT):** States that the distribution of the standardized sample mean approaches a standard Normal distribution $\mathcal{N}(0, 1)$ as the sample size grows large, *regardless* of the underlying population's probability distribution (provided the variance is finite):
  $$
  Z = \frac{\bar{X}_n - \mu}{\sigma / \sqrt{n}} \xrightarrow{d} \mathcal{N}(0, 1)
  $$

> **Key Distinction:** 
> * The **Law of Large Numbers** concerns the *convergence of the value* of an average.
> * The **Central Limit Theorem** concerns the *distribution shape* of sample averages across repeated sampling, enabling confidence intervals and hypothesis testing even for non-normal data.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(202)
flips = rng.binomial(1, 0.3, size=5000)
running_mean = np.cumsum(flips) / np.arange(1, len(flips)+1)
plt.figure(figsize=(8, 4))
plt.plot(running_mean, label="Running sample mean")
plt.axhline(0.3, linestyle="--", label="True probability")
plt.xlabel("Number of trials")
plt.ylabel("Proportion of successes")
plt.title("Law of large numbers: running proportion")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
