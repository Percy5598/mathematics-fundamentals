### Expectation, Variance, and Covariance

* **Expected Value ($E[X]$):** The probability-weighted average of all possible values a random variable can take, representing its theoretical center or long-run average:
  $$
  E[X] = \sum_x x \, P(X = x) \quad \text{(discrete)} \quad \text{or} \quad \int_{-\infty}^{\infty} x f(x) \, dx \quad \text{(continuous)}
  $$

* **Variance ($\text{Var}(X)$):** Measures the spread or dispersion of a random variable around its mean:
  $$
  \text{Var}(X) = E\left[(X - E[X])^2\right]
  $$

* **Covariance ($\text{Cov}(X, Y)$):** Measures the joint variability of two random variables to determine if they tend to move together:
  * **Positive Covariance:** Indicates that when one variable is above its mean, the other tends to be above its mean as well.
  * **Scale Dependency:** Covariance depends on the units/scale of measurement, which motivates standardized measures like correlation.

```python
import numpy as np

rng = np.random.default_rng(10)
x = rng.normal(loc=5, scale=2, size=5_000)
y = 0.7*x + rng.normal(loc=0, scale=1, size=5_000)
print("Sample mean x:", x.mean())
print("Sample variance x:", x.var(ddof=1))
print("Sample covariance:", np.cov(x, y, ddof=1)[0, 1])
print("Correlation:", np.corrcoef(x, y)[0, 1])
```
