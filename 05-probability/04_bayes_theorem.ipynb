### Bayes' Theorem

Bayes' theorem reverses conditional probabilities, allowing us to update our belief about a hypothesis (cause) after observing new evidence:

$$
P(A \mid B) = \frac{P(B \mid A)P(A)}{P(B)}
$$

* **Prior ($P(A)$):** The initial belief or base rate of the cause before considering the new evidence.
* **Likelihood ($P(B \mid A)$):** The probability of observing the evidence $B$ given that the cause $A$ is true.
* **Marginal Probability ($P(B)$):** The total probability of observing the evidence across all possible causes.
* **Posterior ($P(A \mid B)$):** The updated probability of the cause given the observed evidence.

> **Note:** The base rate $P(A)$ critically influences the final posterior probability, a factor that is vital to account for in statistical modeling and machine learning.

```python
# Illustrative screening test example
prevalence = 0.01
sensitivity = 0.95       # P(positive | condition)
specificity = 0.90       # P(negative | no condition)
false_positive_rate = 1 - specificity

p_positive = sensitivity*prevalence + false_positive_rate*(1-prevalence)
p_condition_given_positive = sensitivity*prevalence / p_positive

print("P(positive):", p_positive)
print("P(condition | positive):", p_condition_given_positive)
print("Interpretation: a positive result is not the same as certainty.")
```
