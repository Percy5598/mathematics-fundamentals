### Common Probability Distributions

Different distributions model specific types of processes and data generation mechanisms:

* **Bernoulli Distribution:** Models a single trial with a binary outcome (success/failure, true/false, 0/1).
* **Binomial Distribution:** Models the total number of successes across a fixed number of independent Bernoulli trials (e.g., number of heads in $n$ coin flips).
* **Uniform Distribution:** Models a continuous or discrete process where all outcomes or values within a given interval are equally likely.
* **Normal (Gaussian) Distribution:** A symmetric, bell-shaped distribution characterized by its mean and variance, widely used due to the Central Limit Theorem.
* **Poisson Distribution:** Models the count of discrete events occurring over a fixed interval of time or space at a constant average rate (e.g., customer arrivals).

> **Selection Rule:** Always choose a distribution based on the underlying physical process and assumptions of the data, rather than mathematical convenience alone.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(23)
samples = {
    "Bernoulli": rng.binomial(1, 0.3, 5000),
    "Binomial": rng.binomial(10, 0.3, 5000),
    "Uniform": rng.uniform(0, 1, 5000),
    "Normal": rng.normal(0, 1, 5000),
    "Poisson": rng.poisson(3, 5000),
}
fig, axes = plt.subplots(2, 3, figsize=(11, 6))
for ax, (name, values) in zip(axes.flat, samples.items()):
    ax.hist(values, bins=25, density=True)
    ax.set_title(name)
    ax.grid(alpha=0.2)
axes.flat[-1].axis("off")
plt.tight_layout()
plt.show()
```
