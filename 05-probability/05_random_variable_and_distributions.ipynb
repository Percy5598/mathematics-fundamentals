### Random Variables and Distributions

A random variable is a mathematical formalization that maps outcomes from a sample space to numerical values.

* **Discrete Random Variable:** Takes on a countable number of distinct values (e.g., coin flips, counts). Its probability mass function (PMF) assigns probabilities that sum to 1:
  $$
  \sum_{x} P(X = x) = 1
  $$

* **Continuous Random Variable:** Can take on any value within a continuous interval or set of intervals (e.g., height, time, temperature). Its probability density function (PDF) $f(x)$ describes the relative likelihood of values, where the total area under the density curve is 1:
  $$
  \int_{-\infty}^{\infty} f(x) \, dx = 1
  $$

```python
import numpy as np
import matplotlib.pyplot as plt

# Binomial random variable: number of heads in 10 fair coin flips
rng = np.random.default_rng(7)
simulated = rng.binomial(n=10, p=0.5, size=20_000)
values, counts = np.unique(simulated, return_counts=True)
plt.bar(values, counts / len(simulated))
plt.xlabel("Number of heads in 10 flips")
plt.ylabel("Simulated probability")
plt.title("Binomial distribution estimated by simulation")
plt.grid(axis="y", alpha=0.3)
plt.show()
print("Simulated mean:", simulated.mean(), "theoretical mean:", 10*0.5)
```
