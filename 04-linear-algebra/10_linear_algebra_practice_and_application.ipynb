### Application: A Linear Prediction

A simple linear model can be written as $\hat{y} = X\beta$, where $X$ contains features and $\beta$ contains weights.

```python
import numpy as np

# Each row is an observation; columns are two features.
X = np.array([
    [1.0, 2.0],
    [2.0, 1.0],
    [3.0, 4.0],
    [4.0, 3.0]
])
beta = np.array([10.0, 5.0])
predictions = X @ beta

print("Feature matrix shape:", X.shape)
print("Coefficient vector shape:", beta.shape)
print("Predictions:", predictions)

# Small checks
v = np.array([3, 4])
print("Vector norm:", np.linalg.norm(v))
print("Dot product:", np.dot([2, 3], [4, -1]))
A = np.array([[2, 1], [1, -1]], dtype=float)
b = np.array([5, 1], dtype=float)
print("System solution:", np.linalg.solve(A, b))
print("Determinant:", np.linalg.det(np.array([[3, 1], [2, 4]])))
```
