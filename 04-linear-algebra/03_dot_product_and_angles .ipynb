### 3. Dot Products and Angles

The dot product multiplies matching components and adds the results:
$$
a \cdot b = \sum_{i=1}^n a_i b_i
$$

It is also related to the angle between two nonzero vectors:
$$
a \cdot b = \|a\| \|b\| \cos\theta
$$

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 0, -1])

print("NumPy dot product:", np.dot(a, b))
print("Manual dot product:", 1*4 + 2*0 + 3*(-1))

cos_theta = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
theta_degrees = np.degrees(np.arccos(np.clip(cos_theta, -1, 1)))
print("Angle in degrees:", theta_degrees)
```

### Interpretation
- Positive dot product: angle is acute.
- Zero dot product: vectors are orthogonal (perpendicular).
- Negative dot product: angle is obtuse.

Cosine similarity divides the dot product by both vector lengths. It is often used to compare text embeddings and other feature vectors.

```python
from numpy.linalg import norm

def cosine_similarity(a, b):
    denominator = norm(a) * norm(b)
    if denominator == 0:
        raise ValueError("Cosine similarity is undefined for a zero vector.")
    return np.dot(a, b) / denominator

print(cosine_similarity([1, 0, 1], [1, 1, 0]))
print("Orthogonal example:", np.dot([1, 0], [0, 1]))
```
