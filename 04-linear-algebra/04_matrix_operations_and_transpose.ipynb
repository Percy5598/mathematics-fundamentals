### 4. Matrix Operations and Transpose

A matrix is a two-dimensional array. Matrix addition requires equal shapes. Matrix multiplication has a different rule: if A is m × n and B is n × p, then AB is m × p.

The transpose swaps rows and columns. If A is m × n, then Aᵀ is n × m.

```python
import numpy as np

A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[10, 20, 30], [40, 50, 60]])

print("A + B:\\n", A + B)
print("A shape:", A.shape)
print("A transpose shape:", A.T.shape)
print("A transpose:\\n", A.T)
```

### Matrix multiplication versus elementwise multiplication
- `A * B` multiplies matching elements.
- `A @ B` performs matrix multiplication.

These are not interchangeable.

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Elementwise A * B:\\n", A * B)
print("Matrix product A @ B:\\n", A @ B)
```
