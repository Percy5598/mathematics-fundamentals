### 5. Matrix-Vector Multiplication and Linear Transformations

Matrix-vector multiplication combines weighted inputs:
$$
y = Ax
$$

For a matrix $A$ with $m$ rows and $n$ columns, $x$ must have $n$ components and the output $y$ has $m$ components. Each output is a weighted sum of the input values.

```python
import numpy as np

A = np.array([
    [2, 1],
    [1, 3],
    [0, 4]
])
x = np.array([10, 5])
y = A @ x

print("A shape:", A.shape)
print("x shape:", x.shape)
print("y = A @ x:", y)
print("Each row of A defines one weighted sum.")
```

### A geometric transformation
A matrix can stretch, rotate, reflect, or project vectors. For example, the matrix below doubles the horizontal coordinate and halves the vertical coordinate.

```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([[2, 0], [0, 0.5]])
vectors = np.array([[1, 0], [0, 1], [1, 1], [-1, 1]])

plt.figure(figsize=(6, 6))
for v in vectors:
    transformed = A @ v
    plt.arrow(0, 0, v[0], v[1], head_width=0.06, length_includes_head=True, alpha=0.6)
    plt.arrow(0, 0, transformed[0], transformed[1], head_width=0.06, length_includes_head=True, linestyle="--")
plt.axhline(0, linewidth=1); plt.axvline(0, linewidth=1)
plt.xlim(-3, 3); plt.ylim(-2, 2)
plt.gca().set_aspect("equal", adjustable="box")
plt.title("Original vectors (solid) and transformed vectors (dashed)")
plt.grid(True)
plt.show()
```

### ML connection
A linear layer in a neural network computes a matrix-vector or matrix-matrix product, often followed by a bias and nonlinear activation.
