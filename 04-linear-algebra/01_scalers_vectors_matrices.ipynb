### 1. Scalars, Vectors, and Matrices

### Learning goals
- Distinguish scalars, vectors, and matrices.
- Represent simple data with NumPy.
- Connect linear algebra objects to datasets and ML.

### Intuition
- A **scalar** is one number, such as a stock return of 0.03.
- A **vector** is an ordered list of numbers, such as returns for three assets.
- A **matrix** is a rectangular table of numbers, such as rows of observations and columns of features.

In machine learning, one observation may be a vector, while a dataset is often represented as a matrix.

```python
import numpy as np

scalar = 0.03
vector = np.array([0.03, -0.01, 0.025])
matrix = np.array([
    [0.03, 120.0],
    [-0.01, 115.0],
    [0.025, 123.0]
])

print("Scalar:", scalar, "| shape:", np.shape(scalar))
print("Vector:", vector, "| shape:", vector.shape)
print("Matrix:\\n", matrix, "| shape:", matrix.shape)
print("Rows = observations; columns = features in this example.")
```

### Dimensions matter
A vector with 3 elements has shape `(3,)`. A matrix with 3 rows and 2 columns has shape `(3, 2)`. Many programming errors come from incompatible shapes.
