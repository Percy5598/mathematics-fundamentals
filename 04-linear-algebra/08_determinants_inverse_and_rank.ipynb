### 8. Determinants, Inverse Matrices, and Rank

For a $2 \times 2$ matrix:
$$
A = \begin{bmatrix} a & b \\ c & d \end{bmatrix}, \qquad
\det(A) = ad - bc
$$

If the determinant is nonzero, the square matrix is invertible. If it is zero, the matrix is singular.

```python
import numpy as np

A = np.array([[2, 1], [5, 3]], dtype=float)
print("Determinant:", np.linalg.det(A))
print("Inverse:\\n", np.linalg.inv(A))
print("A @ inverse (approximately identity):\\n", A @ np.linalg.inv(A))
print("Rank:", np.linalg.matrix_rank(A))
```

### Numerical caution
Floating-point calculations can produce tiny rounding errors, so a theoretically exact zero determinant may appear as a very small number. Use numerical tolerances rather than testing exact equality with zero.

In practice, matrix rank and condition number can help diagnose redundancy or numerical instability.

```python
B = np.array([[1, 2], [2, 4]], dtype=float)
print("Determinant of singular matrix:", np.linalg.det(B))
print("Rank:", np.linalg.matrix_rank(B))
print("Condition number:", np.linalg.cond(B))
```
