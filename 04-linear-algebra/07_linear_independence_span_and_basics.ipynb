### 7. Linear Independence, Span, and Basis

A vector is a linear combination of other vectors if it can be built by scaling and adding them.

The **span** of a set is all combinations that can be built. A set is **linearly independent** if no vector in it can be formed from the others (equivalently, only the all-zero coefficients produce a zero combination).

A **basis** is a linearly independent set that spans the space.

```python
import numpy as np

independent = np.array([[1, 0], [0, 1]])
dependent = np.array([[1, 2], [2, 4]])

print("Rank of independent rows:", np.linalg.matrix_rank(independent))
print("Rank of dependent rows:", np.linalg.matrix_rank(dependent))
print("A 2D basis needs two linearly independent vectors.")
```

### Why rank matters
Rank is the number of independent rows or columns. It tells you how many independent directions are represented in a matrix.

In regression, perfect multicollinearity occurs when some feature columns are exact linear combinations of others. This can make coefficient estimates non-unique.

```python
X = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [3, 6, 9]
], dtype=float)

print("Shape:", X.shape)
print("Rank:", np.linalg.matrix_rank(X))
print("There are 3 columns, but only 1 independent direction.")
```
