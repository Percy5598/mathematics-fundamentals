### 6. Systems of Linear Equations

A system can be written compactly as:
$$
Ax = b
$$

Example:
$$
2x + y = 5, \qquad x - y = 1
$$
A solution must satisfy both equations at once.

```python
import numpy as np

A = np.array([[2, 1], [1, -1]], dtype=float)
b = np.array([5, 1], dtype=float)
solution = np.linalg.solve(A, b)

print("Solution [x, y]:", solution)
print("Check A @ solution:", A @ solution)
print("Right-hand side b:", b)
```

### When can a unique solution exist?
For a square system, a unique solution exists when A is invertible. More generally, consistency and uniqueness depend on the relationships among the equations and their coefficients.

Avoid calculating an inverse explicitly just to solve a system; use `np.linalg.solve` when the system is square and well-conditioned.

```python
# A system with no unique solution can raise a linear algebra error.
A_singular = np.array([[1, 2], [2, 4]], dtype=float)
b_singular = np.array([3, 6], dtype=float)

print("Rank:", np.linalg.matrix_rank(A_singular))
print("The rows are dependent, so this system does not have a unique solution.")
```
