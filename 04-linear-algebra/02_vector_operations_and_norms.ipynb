### 2. Vector Operations and Norms

Vectors can be added or subtracted element by element if their shapes match. Multiplying a vector by a scalar scales every component.

The Euclidean norm measures vector length:
$$
\|x\|_2 = \sqrt{x_1^2 + x_2^2 + \cdots + x_n^2}
$$

```python
import numpy as np

a = np.array([2, 3])
b = np.array([1, -1])

print("a + b =", a + b)
print("a - b =", a - b)
print("3a =", 3 * a)
print("Euclidean norm of a =", np.linalg.norm(a))
print("Manually =", np.sqrt(2**2 + 3**2))
```

### Why norms matter
Norms measure magnitude or distance. They appear in:
- regularization, such as L2 penalties in regression;
- similarity and distance calculations;
- optimization stopping rules;
- comparing parameter vectors.

The L1 norm is $\|x\|_1 = \sum_i |x_i|$. The L2 norm is the Euclidean length.

```python
x = np.array([3, -4, 12])
print("L1 norm:", np.linalg.norm(x, ord=1))
print("L2 norm:", np.linalg.norm(x))
print("L-infinity norm:", np.linalg.norm(x, ord=np.inf))
```
