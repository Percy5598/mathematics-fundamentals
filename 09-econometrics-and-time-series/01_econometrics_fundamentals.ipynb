### 1. What is Econometrics?

Econometrics bridges economic theory, mathematics, and statistics to analyze real-world economic and financial relationships. 

A simple linear regression model is expressed as:
$$
Y_i = \beta_0 + \beta_1 X_i + \epsilon_i
$$
where:
* $Y_i$: The dependent outcome variable.
* $X_i$: The explanatory (independent) variable.
* $\beta_0$: The intercept.
* $\beta_1$: The slope coefficient.
* $\epsilon_i$: The error term representing unobserved influences.

> **Core Caution:** Observing a statistical relationship in observational data does not automatically prove causation.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
n = 150

advertising = rng.uniform(0, 100, n)
sales = 20 + 2.5 * advertising + rng.normal(0, 25, n)

data = pd.DataFrame({
    "advertising": advertising,
    "sales": sales,
})

print("Dataset Head:")
display(data.head())
print("\nDataset Summary Statistics:")
display(data.describe())

plt.figure(figsize=(8, 4))
plt.scatter(data["advertising"], data["sales"], color='dodgerblue', alpha=0.7, label="Observations")
plt.xlabel("Advertising Expenditure ($)")
plt.ylabel("Sales ($)")
plt.title("Illustrative Economic Relationship: Advertising vs. Sales")
plt.legend()
plt.grid(linestyle=':', alpha=0.6)
plt.show()
```
