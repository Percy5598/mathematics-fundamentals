## Multiple Regression and Endogeneity

### Learning Goals
* Understand how multiple regression estimates conditional relationships holding included variables constant.
* Recognize the threat of **endogeneity** (correlation between explanatory variables and the error term).
* Identify core causes of endogeneity: omitted variables, reverse causality, and measurement error.

---

### 1. Multiple Regression Specification

A multiple linear regression model extends simple regression by incorporating multiple explanatory variables:
$$
Y_i = \beta_0 + \beta_1 X_{1i} + \beta_2 X_{2i} + \epsilon_i
$$
Each slope coefficient measures the partial effect of an explanatory variable on the outcome, holding all other included variables constant.

> **The Endogeneity Problem:** If an explanatory variable is correlated with the error term ($\mathbb{COV}(X, \epsilon) \neq 0$), OLS estimates become biased and inconsistent.

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

rng = np.random.default_rng(8)
n = 500

education = rng.normal(12, 2, n)
experience = rng.uniform(0, 30, n)
wage = (
    8
    + 1.7 * education
    + 0.25 * experience
    + rng.normal(0, 4, n)
)

data = pd.DataFrame({
    "wage": wage,
    "education": education,
    "experience": experience,
})

X = sm.add_constant(data[["education", "experience"]])
result = sm.OLS(data["wage"], X).fit()
print(result.summary())
```
