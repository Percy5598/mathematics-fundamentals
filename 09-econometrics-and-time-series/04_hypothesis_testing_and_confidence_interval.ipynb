### 1. Testing a Regression Coefficient

A standard two-sided test for a regression slope evaluates whether an explanatory variable has a linear association with the outcome:
$$
H_0: \beta_1 = 0 \quad \text{versus} \quad H_1: \beta_1 \neq 0
$$

> **Key Distinction:** A $p$-value measures how incompatible the observed test statistic is with the null hypothesis under the model's assumptions; it is *not* the probability that the null hypothesis is true.

```python
import numpy as np
import statsmodels.api as sm

rng = np.random.default_rng(4)
x = rng.normal(size=200)
y = 1.5 + 0.7 * x + rng.normal(0, 1, size=200)

result = sm.OLS(y, sm.add_constant(x)).fit()
print(result.summary())

print("\n95% Confidence Intervals for Coefficients:")
print(result.conf_int())
```

### Interpretation and Limitations

Confidence intervals quantify estimation uncertainty under the method's assumptions. However, achieving statistical significance ($p < 0.05$) does not automatically guarantee economic importance or a causal effect.
